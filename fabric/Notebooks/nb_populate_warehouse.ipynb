```python
# Get some public data to use for demo purposes
import com.microsoft.spark.fabric
import pandas as pd

warehouse_name = "wh_gold"

try:
    yellow_taxi_data_url = "https://d37ci6vzurychx.cloudfront.net/trip-data/yellow_tripdata_2025-01.parquet"
    df = spark.createDataFrame(pd.read_parquet(yellow_taxi_data_url))
    df.write.mode("overwrite").synapsesql(f"{warehouse_name}.dbo.source_nyc_taxi_yellow_2025_01")

    green_taxi_data_url = "https://d37ci6vzurychx.cloudfront.net/trip-data/green_tripdata_2025-01.parquet"
    df = spark.createDataFrame(pd.read_parquet(green_taxi_data_url))
    df.write.mode("overwrite").synapsesql(f"{warehouse_name}.dbo.source_nyc_taxi_green_2025_01")

    print(f"Source data loaded to {warehouse_name}")
except Exception as ex:
    print(f"An error occured : {ex}")
```
